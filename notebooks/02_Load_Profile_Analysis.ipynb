```python
import pandas as pd
import matplotlib.pyplot as plt
import os

base_path = r"D:\Dataset\Anmol-MP-EV"

processed_path = os.path.join(base_path, "processed_data")
figure_path = os.path.join(base_path, "figures")

print("Project path loaded")
```

```python
daily_profile = pd.read_csv(
    os.path.join(processed_path, "daily_load_profiles.csv")
)

daily_profile.head()
```

```python
# Calculate average load for each 15-minute interval

average_profile = daily_profile.drop(
    columns=["Station", "Date"]
).mean()

average_profile.head()
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    average_profile.index,
    average_profile.values
)

plt.title(
    "Average EV Charging Load Profile (15-minute Interval)"
)

plt.xlabel("Time Interval")
plt.ylabel("Average Charging Load (kW)")

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()


plt.savefig(
    os.path.join(
        figure_path,
        "average_daily_load_profile.png"
    ),
    dpi=300
)

plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    average_profile.index,
    average_profile.values,
    linewidth=2
)

plt.title(
    "Average EV Charging Load Profile (15-minute Interval)",
    fontsize=14
)

plt.xlabel("Time of Day")
plt.ylabel("Average Charging Load (kW)")


# show every 2 hours only
plt.xticks(
    range(0,96,8),
    average_profile.index[::8],
    rotation=45
)

plt.grid(True, alpha=0.3)

plt.tight_layout()


plt.savefig(
    os.path.join(
        figure_path,
        "average_daily_load_profile.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(
    station_features["Station"],
    station_features["Peak"]
)

plt.title("Peak Charging Load Comparison Across EV Stations")
plt.xlabel("Charging Station")
plt.ylabel("Peak Load (kW)")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    r"D:\Dataset\Anmol-MP-EV\figures\station_peak_load_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    station_features["Station"],
    station_features["Load_Factor"]
)

plt.title("Load Factor Comparison Across EV Charging Stations")
plt.xlabel("Charging Station")
plt.ylabel("Load Factor")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    r"D:\Dataset\Anmol-MP-EV\figures\station_load_factor_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
